*  **ESCALAR** tensor de rango 0  
*  **VECTOR**  tensor de rango 1

**EJEMPLOS:**

```python
# SOLUCIN GENERADA CON CHAT GPT
import numpy as np

# Definir matrices A y B
A = np.array([[2, 29],
              [-24, 9],
              [18, -10]])

B = np.array([[1, -18],
              [27, 11],
              [-17, 8]])

# Coeficientes del sistema en forma de matriz
#  [ [2, -5],   [x]   = [a]
#    [1,  4] ]  [y]     [b]
M = np.array([[2, -5],
              [1,  4]])

X = []
Y = []

# Resolver fila por fila
for a_row, b_row in zip(A, B):
    row_X = []
    row_Y = []
    for a, b in zip(a_row, b_row):
        rhs = np.array([a, b])          # lado derecho (A, B)
        sol = np.linalg.solve(M, rhs)   # resuelve [x, y]
        row_X.append(sol[0])
        row_Y.append(sol[1])
    X.append(row_X)
    Y.append(row_Y)

X = np.array(X, dtype=int)
Y = np.array(Y, dtype=int)

print("Matriz X:")
print(X)
print("\nMatriz Y:")
print(Y)
```

#### Probabilidades
Una población tiene una estatura promedio de 170 cm con una desviación estándar de 10 cm. ¿Cuál es la probabilidad de que una persona mida menos de 180 cm?

```python
from scipy.stats import norm

media = 170
desviacion_estandar = 10

probabilidad = norm.cdf(x=180, loc=media, scale=desviacion_estandar)

print(f"La probabilidad solicitada es: {probabilidad:.2f}")
```

### Prueba de hipotesis

Una empresa de Rio Abajo, afirma que la altura promedio de los árboles en su plantación es de 10 metros. Un agrónomo toma una muestra de 30 árboles y quiere saber si la altura promedio de la muestra es significativamente diferente de 10 metros.

Hipótesis:

H_0 (Hipótesis nula): La altura promedio de los árboles es igual a 10 metros (mu=10).

H_a (Hipótesis alternativa): La altura promedio de los árboles es diferente de 10 metros (mu!=10).

Nivel de significancia ( alpha): Se elige un valor de alpha=0.05.

```python
import numpy as np
from scipy import stats

# Datos de la muestra (simulado)
alturas_muestra = np.random.normal(loc=9.5, scale=1.5, size=30)
media_hipotetica = 10

# Realizar la prueba t de una muestra
# La función ttest_1samp() compara la media de la muestra con la media hipotética.
estadistico_t, valor_p = stats.ttest_1samp(a=alturas_muestra, popmean=media_hipotetica)

print(f"Estadístico t: {estadistico_t:.4f}")
print(f"Valor p: {valor_p:.4f}")

# Tomar una decisión basada en el valor p y el nivel de significancia (alpha)
if valor_p < 0.05:
    print("\nEl valor p es menor que 0.05. Se rechaza la hipótesis nula.")
    print("Hay evidencia estadística para decir que la altura promedio de los árboles es diferente de 10 metros.")
else:
    print("\nEl valor p es mayor que 0.05. No se rechaza la hipótesis nula.")
    print("No hay evidencia suficiente para decir que la altura promedio de los árboles es diferente de 10 metros.")
```

Analogía:
Imagina dos bolsas de canicas:

Bolsa 1 (Baja Entropía): Contiene 99 canicas rojas y 1 azul. La incertidumbre es muy baja, es casi seguro que sacarás una canica roja.

Bolsa 2 (Alta Entropía): Contiene 50 canicas rojas y 50 azules. La incertidumbre es máxima, no sabes qué color sacarás.

#### TAREA (5)

```python
import numpy as np

def entropia(probabilidades):
    """
    Calcula la entropía H(X) de una distribución discreta de probabilidades.

    Parámetros:
    probabilidades : lista o array de probabilidades de los eventos (deben sumar 1)

    Retorna:
    entropía H(X)
    """
    probabilidades = np.array(probabilidades)
    # Filtramos probabilidades cero para evitar log(0)
    probabilidades = probabilidades[probabilidades > 0]
    H = -np.sum(probabilidades * np.log2(probabilidades))
    return H

# Ejemplo 1: Bolsa 1 (Baja Entropía)
bolsa1 = [0.99, 0.01]
H1 = entropia(bolsa1)
print(f"Entropía Bolsa 1: {H1:.4f} unid informacion")

# Ejemplo 2: Bolsa 2 (Alta Entropía)
bolsa2 = [0.5, 0.5]
H2 = entropia(bolsa2)
print(f"Entropía Bolsa 2: {H2:.4f} unid informacion")
```

#### **DISTRIBUCIONES**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, bernoulli, binom, poisson, expon, uniform, t, chi2, beta

np.random.seed(42)
N = 50000  # tamaño de muestra

def plot_continua(rv, x, title, samples=None, bins=80):
    plt.figure(figsize=(7,4))
    plt.plot(x, rv.pdf(x), lw=2, label="PDF teórica")
    if samples is not None:
        plt.hist(samples, bins=bins, density=True, alpha=0.4, label=f"Hist ({len(samples)})")
    plt.title(title)
    plt.xlabel("x"); plt.ylabel("densidad")
    plt.legend(); plt.tight_layout()
    plt.show()

def plot_discreta(rv, x, title, samples=None):
    plt.figure(figsize=(7,4))
    plt.stem(x, rv.pmf(x), use_line_collection=True, basefmt=" ")
    if samples is not None:
        vals, freq = np.unique(samples, return_counts=True)
        plt.bar(vals, freq/np.sum(freq), alpha=0.4, width=0.8, label="Hist (freq. rel.)")
        plt.legend()
    plt.title(title)
    plt.xlabel("k"); plt.ylabel("probabilidad")
    plt.tight_layout()
    plt.show()

# 1) Normal (μ, σ)
mu, sigma = 0, 1
x = np.linspace(mu-4*sigma, mu+4*sigma, 1000)
rv_norm = norm(loc=mu, scale=sigma)
s_norm  = rv_norm.rvs(size=N)
plot_continua(rv_norm, x, f"Normal(μ={mu}, σ={sigma})", samples=s_norm)

# 2) Exponencial (λ)
lam = 1.2
rv_exp = expon(scale=1/lam)  # scale = 1/lambda
x = np.linspace(0, 6/lam, 1000)
s_exp = rv_exp.rvs(size=N)
plot_continua(rv_exp, x, f"Exponencial(λ={lam})", samples=s_exp)

# 3) Uniforme(a, b)
a, b = -2, 3
rv_uni = uniform(loc=a, scale=b-a)
x = np.linspace(a-0.5, b+0.5, 800)
s_uni = rv_uni.rvs(size=N)
plot_continua(rv_uni, x, f"Uniforme({a}, {b})", samples=s_uni, bins=60)

# 4) t de Student (ν)
nu = 5
rv_t = t(df=nu)
x = np.linspace(-6, 6, 1000)
s_t = rv_t.rvs(size=N)
plot_continua(rv_t, x, f"t-Student(ν={nu})", samples=s_t)

# 5) Ji-cuadrada (k)
k = 4
rv_chi2 = chi2(df=k)
x = np.linspace(0, 20, 1000)
s_chi2 = rv_chi2.rvs(size=N)
plot_continua(rv_chi2, x, f"Chi-cuadrado(k={k})", samples=s_chi2)

# 6) Beta(α, β)  en (0,1)
alfa, beta_par = 2, 5
rv_beta = beta(a=alfa, b=beta_par)
x = np.linspace(0, 1, 1000)
s_beta = rv_beta.rvs(size=N)
plot_continua(rv_beta, x, f"Beta(α={alfa}, β={beta_par})", samples=s_beta, bins=80)


def plot_discreta(rv, x, title, samples=None):
    plt.figure(figsize=(7,4))
    # ✅ sin use_line_collection (ya no existe en matplotlib recientes)
    plt.stem(x, rv.pmf(x), basefmt=" ", markerfmt="o", linefmt="-")
    if samples is not None:
        vals, freq = np.unique(samples, return_counts=True)
        plt.bar(vals, freq/np.sum(freq), alpha=0.4, width=0.8, label="Hist (freq. rel.)")
        plt.legend()
    plt.title(title)
    plt.xlabel("k"); plt.ylabel("probabilidad")
    plt.tight_layout()
    plt.show()
```
